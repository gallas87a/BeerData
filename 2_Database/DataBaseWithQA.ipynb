```python
import csv
import glob
import os

def beer_to_csv(rows,file_name):
    #add new entry into csv
    with open(file_name, 'a', newline='',encoding='utf-8') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerows(rows)

def create_csv(file_name):
    #create csv with heading
    with open(file_name, 'w', newline='',encoding='utf-8') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['nr', 'bid','brewery','name','kind', 'link','abv','ibu', 'rating','nr_raters','adding_date','text','country','malt','hops','chocolate','german','banana','vanilla','caramel','yeast','fruit','whiskey','rhum','alcohol','gluten','barrel','rye','barley','cocoa','small/lower','big/upper','Altbier','Barleywine','Blonde','Duppel','Quadrupel','Tripel','Cider','Fruit','IPA','Kölsch','Lager','Non-Alcoholic','Pilsner','Porter','Wheat','Sour','Stout','Ale','Other'])

def Open_Csv(name):
    with open(name, "r+", encoding="utf-8") as file:
        reader = csv.reader(file)
        data=[]
        for row in reader:
            data.append(row)
    return data

def Look_Eliminate_Duplicates(in_data):
    seen = set()
    filtered_data = []

    for row in in_data:
        value = row[1]
        if (value not in seen) and value != "bid":
            seen.add(value)
            filtered_data.append(row)
    return filtered_data

def Keyword_Matching(in_beer_data):
    for row in in_beer_data:
        if ("malt" in row[11]) or ("maláta" in row[11]) or ("slad" in row[11]):
            row.append(1) #row no 12
        else:
            row.append(0)
        if ("hops" in row[11]) or ("komló" in row[11]) or ("hmelja" in row[11]):
            row.append(1) #row no 13
        else:
            row.append(0)
        if ("chocolate" in row[11]) or ("csokoládé" in row[11]) or ("csoki" in row[11]) or ("cokolada" in row[11]):
            row.append(1) #row no 14
        else:
            row.append(0)
        if ("German" in row[11]) or ("német" in row[11])or ("german" in row[11]) or ("nemacki" in row[11]):
            row.append(1) #row no 15
        else:
            row.append(0)
        if ("banana" in row[11]) or ("banán" in row[11]) or ("banana" in row[11]):
            row.append(1) #row no 16
        else:
            row.append(0)
        if ("vanilla" in row[11]) or ("vanília" in row[11]) or ("vanila" in row[11]):
            row.append(1) #row no 17
        else:
            row.append(0)
        if ("caramel" in row[11]) or ("karamella" in row[11]) or ("karamel" in row[11]):
            row.append(1) #row no 18
        else:
            row.append(0)
        if ("yeast" in row[11]) or ("élesztő" in row[11]) or ("kvasac" in row[11]):
            row.append(1) #row no 19
        else:
            row.append(0)
        if ("fruit" in row[11]) or ("gyümölcs" in row[11]) or ("berry" in row[11])or ("berries" in row[11]) or ("voce" in row[11]):
            row.append(1) #row no 20
        else:
            row.append(0)
        if ("bourbon" in row[11]) or ("whiskey" in row[11]):
            row.append(1) #row no 21
        else:
            row.append(0)
        if ("rhum" in row[11]) or ("rum" in row[11]):
            row.append(1) #row no 22
        else:
            row.append(0)
        if ("alcohol" in row[11]) or ("alkohol" in row[11]):
            row.append(1) #row no 23
        else:
            row.append(0)
        if ("gluten" in row[11]) or ("glutén" in row[11]):
            row.append(1) #row no 24
        else:
            row.append(0)
        if ("barrel" in row[11]) or ("hordó" in row[11]) or ("bure" in row[11]):
            row.append(1) #row no 25
        else:
            row.append(0)
        if ("rye" in row[11]) or ("rozs" in row[11]) or ("raz" in row[11]):
            row.append(1) #row no 26
        else:
            row.append(0)
        if ("barley" in row[11]) or ("árpa" in row[11]) or ("jecam" in row[11]):
            row.append(1) #row no 27
        else:
            row.append(0)
        if ("cocoa" in row[11]) or ("kakaó" in row[11]) or ("kakao" in row[11]):
            row.append(1) #row no 28
        else:
            row.append(0)
        if ("alacsony" in row[11]) or ("alsó" in row[11]) or ("kicsi" in row[11]) or ("kevés" in row[11]) or ("pici" in row[11]) or ("hideg" in row[11]) or ("mali" in row[11]) or ("hlad" in row[11]):
            row.append(1) #row no 29
        else:
            row.append(0)
        if ("magas" in row[11]) or ("felső" in row[11]) or ("nagy" in row[11]) or ("sok" in row[11]) or ("meleg" in row[11]) or ("mnogo" in row[11]) or ("toplo" in row[11]):
            row.append(1) #row no 30
        else:
            row.append(0)
        if row[1] == "3169618": #Sörművek Marzen is not properly filled out in untappd
            row[6] = "5.0"
        if row[1] == "376508": #Ruski kvas BIP is not properly filled out in untappd
            row[6]= "0.5" 
        if row[1] == "1726040": #Cigla Svetlo Lager is not properly filled out in untappd
            row[6]= "4.7"
        if row[1] == "418107": #Hobbit Beer is not properly filled out in untappd
            row[6]= "4.9"
        if row[1] == "6717723": #Roasters Paradise Kerinici - Indonézia is not properly filled out in untappd
            row[6]= "13.0"

        #beerkind encoding
        beer_kind=0
        if 'Altbier' in row[4] and beer_kind==0:
            row.append(1) #row no 31
            beer_kind=1
        else:
            row.append(0) #row no 31
        if 'Barleywine' in row[4] and beer_kind==0:
            row.append(1) #row no 32
            beer_kind=1
        else:
            row.append(0) #row no 32
        if 'Blonde' in row[4] and beer_kind==0:
            row.append(1) #row no 33
            beer_kind=1
        else:
            row.append(0) #row no 33
        if 'Duppel' in row[4] and beer_kind==0:
            row.append(1) #row no 34
            beer_kind=1
        else:
            row.append(0) #row no 34
        if 'Quadrupel' in row[4] and beer_kind==0:
            row.append(1) #row no 35
            beer_kind=1
        else:
            row.append(0) #row no 35
        if 'Tripel' in row[4] and beer_kind==0:
            row.append(1) #row no 36
            beer_kind=1
        else:
            row.append(0) #row no 36
        if 'Cider' in row[4] and beer_kind==0:
            row.append(1) #row no 37
            beer_kind=1
        else:
            row.append(0) #row no 37
        if 'Fruit' in row[4] and beer_kind==0:
            row.append(1) #row no 38
            beer_kind=1
        else:
            row.append(0) #row no 38
        if 'IPA' in row[4]and beer_kind==0:
            row.append(1) #row no 39
            beer_kind=1
        else:
            row.append(0) #row no 39
        if 'Kölsch' in row[4] and beer_kind==0:
            row.append(1) #row no 40
            beer_kind=1
        else:
            row.append(0) #row no 40
        if 'Lager' in row[4] and beer_kind==0:
            row.append(1) #row no 41
            beer_kind=1
        else:
            row.append(0) #row no 41
        if 'Non-Alcoholic' in row[4] and beer_kind==0:
            row.append(1) #row no 42
            beer_kind=1
        else:
            row.append(0) #row no 42
        if 'Pilsner' in row[4] and beer_kind==0:
            row.append(1) #row no 43
            beer_kind=1
        else:
            row.append(0) #row no 43
        if 'Porter' in row[4] and beer_kind==0:
            row.append(1) #row no 44
            beer_kind=1
        else:
            row.append(0) #row no 44
        if (('Wiesse' in row[4]) or ('Wheat' in row[4])) and beer_kind==0:
            row.append(1) #row no 45
            beer_kind=1
        else:
            row.append(0) #row no 45
        if ('Sour' in row[4]) and beer_kind==0:
            row.append(1) #row no 46
            beer_kind=1
        else:
            row.append(0) #row no 46
        if ('Stout' in row[4]) and beer_kind==0:
            row.append(1) #row no 47
            beer_kind=1
        else:
            row.append(0) #row no 47
        if 'Ale' in row[4] and beer_kind==0:
            row.append(1) #row no 48
            beer_kind=1
        else:
            row.append(0) #row no 48
        if beer_kind == 0:
            row.append(1) #row no 49
        else:
            row.append(0) #row no 49
    return in_beer_data

def Normalize(unnorm_data):
    for row in unnorm_data:
        if (row[6] != "abv") and (row[8] != "rating"):
            row[6] = float(row[6])/1 #normalize ABV with 15 as arbitrary number
            row[8] = float(row[8])/1 #normalize rating with 5 as the maximum rating
        #no else branch
    return unnorm_data

#QA
#1 remove duplications from both csv-s
#remove duplicates from Top50
Top50Data=Open_Csv("../1_Scraper/Top50Data.csv")
Top50Data = Look_Eliminate_Duplicates(Top50Data)
#remove duplicates from AllData
AllData=Open_Csv("../1_Scraper/AllData.csv")
AllData = Look_Eliminate_Duplicates(AllData)
#remove Top50Data from AllData based on row[1]
top_keys = {row[1] for row in Top50Data}
AllData = [row for row in AllData if row[1] not in top_keys]
#add missing beers to ALLData
#add Borsodi: https://untappd.com/b/borsodi-sorgyar-starbev-borsodi/52232
print(len(AllData))
borsodi = ['2356','52232','Borsodi Sörgyár','Borsodi','Lager - Pale','https://untappd.com/b/borsodi-sorgyar-starbev-borsodi/52232','4.5','N/A','2.88','28154','N/A','The flagship of the brewery is a traditional golden coloured light beer.','hungary']
AllData.append(borsodi)

#2 set up 3 csv-s based on the 2 with cutoff / treshold value
treshold=3.85
good_beers = []
rest_beers = []
for row in AllData:
    try:
        value = float(row[8])
        if value > treshold:
            good_beers.append(row)
        else:
            rest_beers.append(row)

    except (ValueError, IndexError):
        pass
print(rest_beers[-1])        

#3 normalization and adding one hot encoding variables and other new variables
datas=[]
data_type=[]

files=["TopBeerData.csv","GoodBeerData.csv","RestBeerData.csv"]
datas = [Top50Data, good_beers, rest_beers]
for i, data_type in enumerate(datas):
    data_type = Keyword_Matching(data_type)
    data_type = Normalize(data_type)
    datas[i] = data_type

#4 export Csvs
#remove all the CSVs from the folder
csv_files = glob.glob("*.csv")
for file in csv_files:
   os.remove(file)
   print(f"Törölve: {file}")
    
for i,data_type in enumerate(datas):
    create_csv(files[i])
    beer_to_csv(data_type, files[i])
    print(f"{files[i]} file created")
```
